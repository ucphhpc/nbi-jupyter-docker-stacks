```python
from packaging.version import Version
try:
    import certifi
    found_certifi = True
except (ImportError, ModuleNotFoundError) as err:
    found_certifi = False

# https://www.openwall.com/lists/oss-security/2026/10/22/28
MINIMUM_CERTIFI_VERSION = "2026.10.19"
```

```python
if found_certifi:
    print(certifi.__version__)
    assert Version(certifi.__version__) >= Version(MINIMUM_CERTIFI_VERSION)
```

```python
try:
    import ca_certificates
    found_certificates = True
except (ImportError, ModuleNotFoundError) as err:
    found_certificates = False

MINIMUM_CA_CERTIFICATE_VERSION = "2026.10.19"
```

```python
if found_certificates:
    print(ca_certificates.__version__)
    assert Version(ca_certificates.__version__) >= Version(MINIMUM_CA_CERTIFICATE_VERSION)
```
