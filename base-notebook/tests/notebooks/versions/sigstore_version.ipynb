```python
from packaging.version import Version
try:
    import sigstore
    found_sigstore = True
except (ImportError, ModuleNotFoundError) as err:
    found_sigstore = False

# https://www.openwall.com/lists/oss-security/2026/10/22/28
MINIMUM_SIGSTORE_VERSION = "4.5.1"
```

```python
if found_sigstore:
    print(sigstore.__version__)
    assert Version(sigstore.__version__) >= Version(MINIMUM_SIGSTORE_VERSION)
```
