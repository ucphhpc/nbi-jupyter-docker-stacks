```python
from packaging.version import Version
try:
    import pip
    found_pip = True
except (ImportError, ModuleNotFoundError) as err:
    found_pip = False

# https://www.openwall.com/lists/oss-security/2026/10/22/28
MINIMUM_PIP_VERSION = "26.2.2"
```

```python
if found_pip:
    print(pip.__version__)
    assert Version(pip.__version__) >= Version(MINIMUM_PIP_VERSION)
```
