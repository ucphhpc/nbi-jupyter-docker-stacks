```python
from packaging.version import Version
import os
import subprocess

# https://www.openwall.com/lists/oss-security/2026/10/22/28
CONDA_MINIMUM_VERSION = "26.9.4"
conda_command = ["conda", "--version"]
```

```python
result = subprocess.run(conda_command, check=True, capture_output=True)
assert result.returncode == 0
output = result.stdout.decode("utf-8")
conda_installed_version = output.split()[1].strip()
```

```python
print(conda_installed_version)
assert Version(conda_installed_version) >= Version(CONDA_MINIMUM_VERSION)
```
